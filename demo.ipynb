# Diabetic Retinopathy Grading — Live Demo
Enhanced LSD-HybridViT with ordinal grading, Monte-Carlo-dropout uncertainty and Grad-CAM.

**How to use:** run the cells from top to bottom (`Runtime → Run all`). When asked, upload a
colour fundus photograph (PNG/JPG). The model predicts the DR grade (0–4), shows a Grad-CAM
heatmap, estimates its own uncertainty, and recommends whether to refer the patient.

*Research prototype — not a medical device.*
